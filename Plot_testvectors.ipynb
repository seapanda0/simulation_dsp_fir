```python
import matplotlib.pyplot as plt

# Read first file (hex values)
data1 = []
with open("testvector_input.txt", "r") as f:
    for line in f:
        data1.append(int(line.strip(), 16))

# Read second file (hex values)
data2 = []
with open("testvector_output.txt", "r") as f:
    for line in f:
        data2.append(int(line.strip(), 16))

# Plot both
plt.plot(data1, label="Input")
plt.plot(data2, label="Output", linestyle='--')

plt.title("Comparison of Output vs Test Vector")
plt.xlabel("Index")
plt.ylabel("Value (decimal)")
plt.legend()
plt.grid(True)

plt.show()
```
